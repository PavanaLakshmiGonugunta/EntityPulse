```python
import json
from IPython.display import display, HTML

# 1️⃣ Load your JSON file
with open("finentity.json", "r", encoding="utf-8") as f:
    data = json.load(f)

# 2️⃣ Extract only unique sentences
unique_sentences = list({item["content"] for item in data})

# 3️⃣ Convert into list of dicts with key "content"
sentences_only = [{"content": s} for s in unique_sentences]

# 4️⃣ Save as new JSON file
output_filename = "sentences_only.json"
with open(output_filename, "w", encoding="utf-8") as f:
    json.dump(sentences_only, f, ensure_ascii=False, indent=4)

# 5️⃣ Create a download link (works in Jupyter/Colab)
display(HTML(f"""
<a href="{output_filename}" download="{output_filename}" target="_blank">
📥 Click here to download {output_filename}
</a>
"""))
```

```python
import json
import nltk
from nltk.tokenize import word_tokenize

# Download tokenizer if not available
nltk.download('punkt', quiet=True)

# === Step 1: Load your sentences ===
with open("sentences_only.txt", "r", encoding="utf-8") as f:
    sentences = [line.strip() for line in f if line.strip()]

# === Step 2: Define known aspect keywords ===
# You can customize or expand this list
aspect_keywords = [
    "battery", "screen", "camera", "performance", "display",
    "sound", "speaker", "design", "price", "storage", "processor"
]

# === Step 3: Create labeled dataset ===
dataset = []

for sentence in sentences:
    tokens = word_tokenize(sentence)
    labels = []

    for i, token in enumerate(tokens):
        token_lower = token.lower()

        if token_lower in aspect_keywords:
            # Label single-word aspect as B-ASP
            labels.append("B-ASP")
        else:
            labels.append("O")

    dataset.append({
        "sentence": sentence,
        "tokens": tokens,
        "labels": labels
    })

# === Step 4: Save dataset as JSON file ===
output_file = "aspect_dataset.json"
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(dataset, f, indent=4, ensure_ascii=False)

print(f"✅ Dataset created and saved as '{output_file}' with {len(dataset)} sentences.")
```
